NumPy – Problem 1: Student Marks Analysis

```python
import numpy as np

marks = np.array([85, 72, 91, 68, 77, 95, 63, 88, 74, 81])
print("Original Array")
print(marks)

sum = np.sum(marks)
print("Sum of marks")
print(sum)

avg = np.mean(marks)
print("Average marks:")
print(avg)

min = np.min(marks)
print("min mark")
print(min)

max = np.max(marks)
print("max mark")
print(max)

print("Display the marks greater than 75")
for i in range(len(marks)):
    if marks[i] > 75:
        print(marks[i])
```

NumPy – Problem 2: Temperature Analysis

```python
import numpy as np
temperatures = np.array([32, 34, 30, 33, 35, 36, 34])
print("original Temperature")
print(temperatures)
```

```python
print("the average temperature")
print(np.mean(temperatures))
```

```python
print("Find the highest and lowest temperature")
print(np.max(temperatures))
print(np.min(temperatures))
```

```python
for i in range(len(temperatures)):
    if temperatures[i] > 30:
        print("the days where the temperature was above 30°C", i+1)
```

```python
updated_temperatures = temperatures + 2

print("\nUpdated temperatures after increasing by 2°C:")
print(updated_temperatures)
```

Pandas – Problem 3: Student Performance DataFrame

```python
import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Charan", "Divya", "Esha", "Farhan", "Gokul", "Hari"],
    "Department": ["CSE", "ECE", "CSE", "IT", "EEE", "CSE", "IT", "ECE"],
    "Marks": [85, 72, 90, 68, 78, 95, 74, 82],
    "Attendance": [92, 78, 85, 75, 88, 95, 72, 81]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print("\nFirst 5 Students:")
print(df.head(5))

average_marks = df["Marks"].mean()
print("\nAverage Marks:", average_marks)

print("\nStudents who scored more than 75:")
print(df[df["Marks"] > 75])

print("\nStudents whose attendance is below 80%:")
print(df[df["Attendance"] < 80])

print("\nStudents sorted by marks:")
print(df.sort_values(by="Marks"))
```

```python
import pandas as pd

data = {
    "Product Name": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse", "Tablet"],
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Accessories", "Electronics"],
    "Price": [50000, 25000, 2000, 1500, 800, 18000],
    "Quantity Sold": [25, 60, 80, 45, 70, 35]
}

df = pd.DataFrame(data)

print("Product Data:")
print(df)

df["Total Sales"] = df["Price"] * df["Quantity Sold"]

print("\nTotal Sales for Each Product:")
print(df[["Product Name", "Total Sales"]])

highest_sales_product = df.loc[df["Total Sales"].idxmax()]

print("\nProduct with Highest Sales:")
print(highest_sales_product)

average_price = df["Price"].mean()

print("\nAverage Product Price:", average_price)

print("\nProducts with Quantity Sold Greater Than 50:")
print(df[df["Quantity Sold"] > 50])

print("\nProducts Sorted by Total Sales:")
print(df.sort_values(by="Total Sales", ascending=False))
```
