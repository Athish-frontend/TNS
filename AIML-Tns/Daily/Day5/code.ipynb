```python
import numpy as np
Marks = np.array([[67,43,54],[89,38,86],[30,55,67],[89,87,55],[56,86,28],[89,27,58]])
print(Marks)
```

```python

print("\nAverage Subject Mark")

average_sub = np.mean(Marks, axis=0)
print(average_sub)

print("\nAverage Student Mark")

average_stu = np.mean(Marks, axis=1)
print(average_stu)
```

```python
for i in range(len(average_stu)):

    if average_stu[i] > 70:
        print("Student", i, "with average of", average_stu[i])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

month = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

sales = [
    120, 150, 180, 220,
    300, 380, 450, 430,
    350, 280, 200, 160
]

df = pd.DataFrame({
    "Month": month,
    "Sales": sales
})

print(df)
```

```python
plt.plot(month, sales, marker="o")

plt.xlabel("Month")
plt.ylabel("Ice Cream Sales")
plt.title("Ice Cream Sales - Line Plot")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.scatter(range(1, 13), sales)

plt.xlabel("Month")
plt.ylabel("Ice Cream Sales")
plt.title("Ice Cream Sales - Scatter Plot")

plt.show()
```

```python
plt.bar(month, sales)

plt.xlabel("Month")
plt.ylabel("Ice Cream Sales")
plt.title("Ice Cream Sales - Bar Graph")

plt.xticks(rotation=45)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

print(tips.head())

numeric_data = tips.select_dtypes(include="number")

correlation = numeric_data.corr()

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap - Tips Dataset")
plt.show()

```
