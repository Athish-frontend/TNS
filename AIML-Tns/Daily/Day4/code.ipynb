```python
marks = [77,67,45,90,87]
print("Marks:" ,marks)
for i in range(len(marks)):
    for j in range(0, len(marks) - 1 - i):
        if marks[j] > marks[j+1]:
            marks[j],marks[j+1]  = marks[j+1],marks[j]
print("Sorted marks",marks)
```

```python
arr = [10, 20, 10, 30, 20, 10]

count = {}

for num in arr:
    if num in count:
        count[num] = count[num] + 1
    else:
        count[num] = 1

for n in count:
    print(n, count[n])

print(count)
```
