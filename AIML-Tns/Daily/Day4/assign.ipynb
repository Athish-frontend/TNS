```python
arr2 = [10, 5, 0, 7, 8, 0, 13, 4]

even = 0
odd = 0

print(arr2.count(0))

for i in range(len(arr2)):

    if arr2[i] % 2 == 0:
        even += 1
    else:
        odd += 1

print(odd)
print(even)
```

```python
arr = [1, 2, 3, 5, 6, 7]

n = 7

expected = n * (n + 1) // 2

actual = 0

for i in range(len(arr)):
    actual = actual + arr[i]

missing = expected - actual

print("Missing Number:", missing)
```

```python
arr = [2, 5, 2, 8, 5, 2, 3, 5, 2]

max_frequency = 0
most_frequent = arr[0]

for i in range(len(arr)):
    count = 0

    for j in range(len(arr)):
        if arr[i] == arr[j]:
            count = count + 1

    if count > max_frequency:
        max_frequency = count
        most_frequent = arr[i]

print("Most Frequent Element:", most_frequent)
print("Frequency:", max_frequency)
```

```python
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

current_sum = arr[0]
max_sum = arr[0]

start = 0
best_start = 0
best_end = 0

for i in range(1, len(arr)):

    if current_sum + arr[i] < arr[i]:
        current_sum = arr[i]
        start = i
    else:
        current_sum = current_sum + arr[i]

    if current_sum > max_sum:
        max_sum = current_sum
        best_start = start
        best_end = i

print("Maximum Subarray Sum:", max_sum)

print("Subarray:", end=" ")
for i in range(best_start, best_end + 1):
    print(arr[i], end=" ")
```

```python
arr1 = [1, 2, 2, 3, 4]
arr2 = [4, 2, 1, 2, 3]

if len(arr1) != len(arr2):
    print("Arrays are Not Equal")
else:
    equal = True

    for i in range(len(arr1)):

        count1 = 0
        count2 = 0

        for j in range(len(arr1)):
            if arr1[i] == arr1[j]:
                count1 = count1 + 1

        for j in range(len(arr2)):
            if arr1[i] == arr2[j]:
                count2 = count2 + 1

        if count1 != count2:
            equal = False
            break

    if equal:
        print("Arrays are Equal")
    else:
        print("Arrays are Not Equal")
```
